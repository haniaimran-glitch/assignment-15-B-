```python
df=pd.read_csv('student_tea_snacks.csv')
```

Q1: What is the most preferred tea type?

```python
pref_tea=df['tea_type'].value_counts()
print(pref_tea)
if pref_tea['Green']>pref_tea['Doodh Patti'] and pref_tea['Green']>pref_tea['Black']:
    print('Green tea is preferred the most')
elif pref_tea['Doodh Patti']>pref_tea['Green'] and pref_tea['Doodh Patti']>pref_tea['Black']:
    print('Doodh Patti is most preferred')
else:
    print('Black tea is most preferred')
```

Q2: What is the average number of cups per gender?

```python
df.groupby('gender')['cups_per_day'].mean().round(2)
```

Q3: Make a bar chart of snack choices.

```python
import matplotlib.pyplot as plt
```

```python
choices=df['snack_choice'].value_counts()
```

```python
categories = choices.index
values = choices.values
plt.bar(categories, values, color='violet', width=0.5)
plt.xlabel('Snack Choices')
plt.ylabel('Frequency')
plt.title('Student Snack Choices')
plt.show()
```

Q4: Compare tea preference by age group.
